```python
#lab-2
#1adding missing values
import pandas as pd
import numpy as np
df=pd.DataFrame({'Age':[25,30,np.nan,40,35],
   'Department':['HR','Finance',np.nan,'IT','teacher']})
print("Original Dataset(with missing values)")
print(df)
df['Age']=df['Age'].fillna(df['Age'].mean())
df['Department']=df['Department'].fillna(df['Department'].mode()[0])
print(df)
```

```python
#2forward/Backward filling
import pandas as pd
import numpy as np
df=pd.DataFrame({'Age':[25,30,np.nan,40,35],
   'Department':['HR','Finance',np.nan,'IT','teacher']})
print("Original Dataset(with missing values)")

print(df)
df_ffill=df.copy()
df_ffill.ffill(inplace=True)
print(df_ffill)

df_bfill=df.copy()
df_bfill.bfill(inplace=True)
print(df_bfill)
```

```python
#3Deletion(drop row or coloumn)
import pandas as pd
import numpy as np
df=pd.DataFrame({'Age':[25,30,np.nan,40,35],
   'Department':['HR','Finance','IT','IT','teacher']})
print("Original Dataset(with missing values)")
print(df)

df_drop_rows=df.dropna()
print("After dropping rows:\n", df_drop_rows)

df_drop_cols=df.dropna(axis=1)
print("After dropping columns:\n", df_drop_cols)
```

```python
#3.b Remove Duplicates
import pandas as pd
df=pd.DataFrame({
    'ID': [1,2,2,3,4,4],
    'Name':['Alice','Bob','Bob','Charlie','David','David'],
    'Age':[25,30,30,35,40,40]
        })
print("Original Data:\n",df)

df_exact=df.drop_duplicates()
print("\nAfter Exact Match Removal:\n", df_exact)
```

```python
3.C#Subset-Based removal(Remove duplicates based on one parameter)
import pandas as pd
df=pd.DataFrame({
    'ID': [1,2,2,3,4,4,5],
    'Name':['Alice','Bob','Bob','Charlie','David','David','David'],
    'Age':[25,30,30,35,40,40,40]
        })
print("Original Data:\n",df)

df_subset_id=df.drop_duplicates(subset=['ID'])
print("\nAfter Subset-Based Removal(ID):\n",df_subset_id)

df_subset_names=df.drop_duplicates(subset=['Name'])
print("\n After Subset-Based Removal(name):\n",df_subset_names)
```

```python
#Removing  Duplicates(Correcting Inconsistent Format)
#3.B.A Date Standardization (ISO 8601 format)

df=pd.DataFrame({
    'Date':['2025-01-05','05/01/26','Jan 5,2026', '2025.01.05']
})
print("Original Data:\n", df)
df['Date']=pd.to_datetime(df['Date'],errors='coerce').dt.strftime('%Y-%m-%d')
print(df)
```

```python
#Removing  Duplicates(Correcting Inconsistent Format)
#3.b.b Case Normalization
df=pd. DataFrame({
    'Name':['MOUNI','mouni','Mounitha','MoUnI']
})
print("Original Data:\n",df)

df['Name_lower']=df['Name'].str.lower()

df['Name_upper']=df['Name'].str.upper()

print("\n",df)
```
